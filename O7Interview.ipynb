```python
import pandas as pd
```

```python
dataset = pd.read_csv('/content/Placement_Data_Full_Class.csv')
```

```python
SSC_percentage_mean = dataset['ssc_p'].mean()
print("SSC_percentage_mean: ",SSC_percentage_mean)
HSC_percentage_mean = dataset['hsc_p'].mean()
print("HSC_percentage_mean: ",HSC_percentage_mean)
degree_percentage_mean = dataset['degree_p'].mean()
print("degree_percentage_mean: ",degree_percentage_mean)
etest_percentage_mean = dataset['etest_p'].mean()
print("etest_percentage_mean: ",etest_percentage_mean)
```

```python
SSC_percentage_mode = dataset['ssc_p'].mode()
print("SSC_percentage_mode: ",SSC_percentage_mode)
HSC_percentage_mode = dataset['hsc_p'].mode()
print("HSC_percentage_mode: ",HSC_percentage_mode)
degree_percentage_mode = dataset['degree_p'].mode()
print("degree_percentage_mode: ",degree_percentage_mode)
etest_percentage_mode = dataset['etest_p'].mode()
print("etest_percentage_mode: ",etest_percentage_mode)
```

```python
SSC_percentage_median = dataset['ssc_p'].median()
print("SSC_percentage_median: ",SSC_percentage_median)
HSC_percentage_median = dataset['hsc_p'].median()
print("HSC_percentage_median: ",HSC_percentage_median)
degree_percentage_median = dataset['degree_p'].median()
print("degree_percentage_mode: ",degree_percentage_median)
etest_percentage_median = dataset['etest_p'].median()
print("etest_percentage_mode: ",etest_percentage_median)
```
